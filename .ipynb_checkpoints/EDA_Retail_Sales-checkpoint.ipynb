```python
age_groups = pd.cut(df['Customer_Age'],bins=[0,18,30,45,60,120],
labels=['Below 18','18-30','31-45','45-60','Above 60'])

age_groups.value_counts().sort_index()
```

```python
import matplotlib.pyplot as plt 
age_groups.value_counts().sort_index().plot(kind='bar',figsize=(8, 5), title='Customer Age Group Distribution') 
plt.xlabel('Age Group')
plt.ylabel('Numbers of Customers')
plt.xticks(rotation=0)
plt.show()
```

### Observation
most customers belong to the 31-45 age group, followed by the 45-60 and 18-30 . Customers below 18 and above 60 are very few

```python
correlation = df[['Sales_Amount','Cost_Amount','Profit','Quantity','Unit_Price','Customer_Age','Customer_Rating','Delivery_Days', 'Inventory_Level']].corr()
correlation.round(2)
```

```python
df.columns.tolist()
```

```python
import seaborn as sns 
import matplotlib.pyplot as plt
plt.figure(figsize=(12,8))
sns.heatmap(correlation,annot=True, cmap='coolwarm',fmt='.2f')
plt.title ('Correlation Heatmap')
plt.show()
```

### Observation
Sales Amount and Cose Amount have a very strong positive relationship. Profit is also strongly related to sales amount. Other variables have a very weak relationship with each other.
